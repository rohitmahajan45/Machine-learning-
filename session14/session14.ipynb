```python
# K-means is an unsupervised Machine learning algorithm used for clustering
# It groups similar data points together into k distinct cluster based on their features
# The goal is to minimise the within cluster varience (i. e. keep points in the same cluster
# as close as possible to the cluster center)
# 1)Clusters:-Groups of similar data points
# Centroid:-The mean(center of a cluster).
# K:-Number of cluster
# Distance metric:-Usually Euclidean distance is used to assign points to the nearest
# cluster
# Iterative process:-Repeats until the centroid stablise
```
