```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error
```

```python
df=pd.read_csv("Advertising.csv")
```

```python
sns.heatmap(df.corr(), annot=True)
```

```python
df.drop("Unnamed: 0", axis=1, inplace=True)
```

```python
df.head()
```

```python
# separate input and output features
```

```python
X=df[['TV']] # independent variable
Y=df['Sales'] # dependent variable
```

```python
# peform splitting of training and testing data
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.3, random_state=42)
```

```python
# create an object of linear regression
```

```python
lr=LinearRegression()
```

```python
# this object is going to call OLS(Ordinary Least Squares) which is used to give best thetal and theta0 value
```

```python
lr.fit(x_train, y_train)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
# now next we have to calculate predicted value on training data
```

```python
y_pred=lr.predict(x_train)
```

```python
plt.figure()
plt.scatter(x_train, y_train)
plt.plot(x_train, y_pred, color='red')
plt.show()
```

```python
# Now perform prediction on testing data
```

```python
y_pred=lr.predict(x_test)
```

```python
# Model evaluation
```

```python
mse=mean_squared_error(y_pred, y_test)
rmse=root_mean_squared_error(y_pred, y_test)
mae=mean_absolute_error(y_pred, y_test)
```

```python
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
print("Mean Absolute Error:", mae)
```
